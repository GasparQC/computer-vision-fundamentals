```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Cargar imágenes
img1 = cv2.imread("/content/descarga (2).jpg")   # imagen color
img2 = cv2.imread("/content/descarga (1).jpg")   # imagen en blanco y negro / ya en tonos de gris visualmente

# Conversión
img1_rgb = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
img1_gray = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
img2_gray = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)

# Dimensiones
print("=== DIMENSIONES ===")
print("Imagen color (img1):", img1_rgb.shape)
print("Imagen color en grises:", img1_gray.shape)
print("Imagen blanco y negro (img2):", img2_gray.shape)

# Pixel central img1
h1, w1 = img1_gray.shape
x1, y1 = w1 // 2, h1 // 2
pixel_central_rgb = img1_rgb[y1, x1]
pixel_central_gray = img1_gray[y1, x1]

print("\n=== PÍXEL CENTRAL DE img1 ===")
print("Ubicación (x, y):", (x1, y1))
print("Valor RGB:", pixel_central_rgb)
print("Valor en grises:", pixel_central_gray)

# Separar canales
r, g, b = cv2.split(img1_rgb)
ceros = np.zeros_like(r)

canal_rojo = cv2.merge([r, ceros, ceros])
canal_verde = cv2.merge([ceros, g, ceros])
canal_azul = cv2.merge([ceros, ceros, b])

# Mostrar imágenes
plt.figure(figsize=(16,8))

plt.subplot(2,3,1)
plt.imshow(img1_rgb)
plt.title("Imagen color original")
plt.axis("off")

plt.subplot(2,3,2)
plt.imshow(img1_gray, cmap="gray")
plt.title("Imagen color -> escala de grises")
plt.axis("off")

plt.subplot(2,3,3)
plt.imshow(img2_gray, cmap="gray")
plt.title("Imagen blanco y negro")
plt.axis("off")

plt.subplot(2,3,4)
plt.imshow(canal_rojo)
plt.title("Canal rojo")
plt.axis("off")

plt.subplot(2,3,5)
plt.imshow(canal_verde)
plt.title("Canal verde")
plt.axis("off")

plt.subplot(2,3,6)
plt.imshow(canal_azul)
plt.title("Canal azul")
plt.axis("off")

plt.tight_layout()
plt.show()
```

Qué diferencias hay entre la imagen color y la imagen en blanco y negro? En qué se parecen? Hay diferencias en tamaño?


La imagen a color tiene 3 canales: rojo, verde y azul (RGB).
La imagen en blanco y negro o escala de grises tiene 1 solo canal, que representa intensidades de luz entre negro y blanco.

Diferencias:

La imagen color guarda más información visual porque distingue colores.
La imagen en blanco y negro simplifica la información y solo muestra luminosidad.
En este caso también hay diferencia de tamaño:
Imagen color: (194, 259, 3)
Imagen blanco y negro: (168, 300)

Eso significa que no solo cambia la cantidad de canales, sino también la resolución.

Similitudes:

Ambas son matrices de píxeles.
Ambas representan una imagen digital.
En las dos se puede analizar el valor de cada píxel.

¿Qué canal resalta más el objeto principal de la imagen color? ¿Por qué?

Según la visualización de los canales, el canal verde (G) es el que mejor resalta el objeto principal.

¿Por qué?
Porque en este canal se observan mejor los detalles del contorno, la camiseta y la separación respecto del fondo. Además, en imágenes digitales el canal verde suele conservar mucha información de luminancia y contraste, por eso frecuentemente se ve más definido.

¿Puedo convertir una imagen de blanco y negro a color?


Sí, pero con una aclaración importante.
Se puede convertir técnicamente una imagen de blanco y negro a una imagen de 3 canales, por ejemplo pasándola de gris a RGB. Sin embargo, eso no recupera los colores originales.
Lo que se obtiene es una imagen “a color” en formato RGB, pero con los tres canales iguales, por lo que sigue viéndose en grises.

Puedo convertir una imagen de color a blanco y negro?

Sí.
Se puede convertir una imagen de color a escala de grises usando una función como:

cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)

Al hacerlo, la imagen pasa de tener 3 canales a tener 1 solo canal.

Reconvertí la imagen a color en escala de grises. ¿Se ven iguales?

No se ven exactamente iguales.

La imagen color original contiene información de color real.
La imagen convertida a escala de grises conserva formas, luces y sombras, pero pierde la información cromática.
Por eso el contenido sigue siendo reconocible, pero visualmente no es igual.

¿Qué valor tiene el píxel central en RGB y qué valor tiene en escala de grises?

Para la imagen color (img1), el píxel central está en la posición:

(129, 97)

Sus valores son:

RGB: [65, 22, 65]
Escala de grises: 40

Esto muestra que el mismo píxel cambia su representación:

en RGB se expresa con 3 valores,
en grises se resume a un solo valor de intensidad.
