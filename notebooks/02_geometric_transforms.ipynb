```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Cargar imagen
img = cv2.imread("/content/descarga (2).jpg")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
# Obtener dimensiones
filas, cols, ch = img.shape
print("filas: ", filas)
print("columnas: ", cols)
print("canales: ", ch)
```

```python
# 1. Desplazamiento
M_despl = np.float32([[1, 0, 50], [0, 1, 30]])
img_despl = cv2.warpAffine(img_rgb, M_despl, (cols, filas))
img_despl
```

```python
# 2. Rotación
M_rot = cv2.getRotationMatrix2D((cols/2, filas/2), 45, 1)
img_rot = cv2.warpAffine(img_rgb, M_rot, (cols, filas))
img_rot
```

```python
# 3. Cambio de tamaño
img_tamaño = cv2.resize(img_rgb, None, fx=1.5, fy=1.5)
img_tamaño
```

```python
# 4. Transformación afín
pts1 = np.float32([[50, 50], [200, 50], [50, 150]])
pts2 = np.float32([[10, 100], [200, 50], [100, 180]])
M_afin = cv2.getAffineTransform(pts1, pts2)
img_afin = cv2.warpAffine(img_rgb, M_afin, (cols, filas))
img_afin
```

```python
# 5. Transformación proyectiva
pts1_proj = np.float32([[0, 0], [cols-1, 0], [0, filas-1], [cols-1, filas-1]])
pts2_proj = np.float32([[50, 20], [cols-50, 0], [0, filas-1], [cols-30, filas-30]])
M_proj = cv2.getPerspectiveTransform(pts1_proj, pts2_proj)
img_proj = cv2.warpPerspective(img_rgb, M_proj, (cols, filas))
img_proj
```

## Respuestas

### 1) ¿Qué cambia cuando la imagen se desplaza?

Cuando la imagen se desplaza, cambia su **posición** dentro del plano.  
No cambia su forma, ni su tamaño, ni su orientación.  
El objeto sigue siendo el mismo, solo se mueve dentro de la imagen.

---

### 2) ¿Qué diferencias se observan entre la imagen original y la rotada?

En la imagen rotada cambia la **orientación**.  
El contenido es el mismo, pero aparece girado.  
También pueden aparecer espacios vacíos o partes recortadas en los bordes.

---

### 3) ¿Qué sucede al modificar el tamaño de la imagen?

La imagen se ve **más grande o más chica**.  
Si se agranda mucho, puede perder calidad o verse pixelada.  
Si se reduce, pierde detalles.

---

### 4) ¿Qué sucede con la forma de la imagen al deformarla?

La forma deja de conservarse exactamente.  
La imagen puede verse inclinada, estirada o torcida, cambiando sus proporciones originales.

---

### 5) ¿Qué efectos visuales produce la deformación o warping?

La deformación genera un efecto visual más evidente.  
Puede dar sensación de inclinación o perspectiva, como si la imagen estuviera vista desde otro ángulo.

---

### 6) ¿Qué diferencias hay entre una transformación afín y una proyectiva?

La transformación **afín** permite mover, rotar, escalar o inclinar la imagen, pero mantiene el paralelismo de las líneas.  

La transformación **proyectiva** modifica la perspectiva, por lo que las líneas paralelas pueden dejar de ser paralelas.

---

### 7) ¿En qué casos se conserva la información original y en cuáles se modifica visualmente?

Se conserva más en:
- desplazamiento  
- rotación  
- cambio de tamaño  

Se modifica más en:
- transformación afín  
- transformación proyectiva  

---

### 8) ¿Qué transformación te resultó más útil o interesante? ¿Por qué?

La transformación proyectiva fue la más interesante porque genera un efecto de perspectiva más marcado.  
Permite ver cómo una imagen puede cambiar el punto de vista y parecer más tridimensional.
